```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpt

# Load the data
data = pd.read_csv('https://raw.githubusercontent.com/Andrew-Hamacher/sipe-data/refs/heads/main/250210_LICK_ses3.csv')
data = data.to_numpy()

# plotting capacitance over time
plt.subplot(3,1,1)
plt.plot((data[:,0]/60), data[:, 1])
plt.ylim(0,200)
plt.xlabel('Time(min)')
plt.title('Cage 1 (1M Ohm Resistor)')
plt.subplot(3,1,2)
plt.plot((data[:,0]/60), data[:, 2])
plt.ylim(0, 500)
plt.xlabel('Time(min)')
plt.title('Cage 2 (4.7M Ohm resistor)')
plt.subplot(3,1,3)
plt.plot((data[:,0]/60), data[:, 3])
plt.ylim(0, 200)
plt.xlabel('Time(min)')
plt.title('Cage 3 2M Ohm resistor')
plt.tight_layout()
#plt.show()

```

```python
data = pd.read_csv('https://raw.githubusercontent.com/Andrew-Hamacher/sipe-data/refs/heads/main/ca_pilot.csv')
data = data.to_numpy()

#running a loop that plots every spout
fig, ax = plt.subplots(22,1, figsize=(10, 20))
for i in range(22):
    ax[i].plot(data[:,0], data[:,(i+1)])
    #ax[i].set_xlim(0,2.5).    #Setting limits for every graph.
    #ax[i].set_ylim(0, 50)
    ax[i].set_title('Channel ' + str(i+1))
plt.tight_layout()
plt.show()

```

```python
#Importing first week of lick data
data1 = pd.read_csv('https://raw.githubusercontent.com/Andrew-Hamacher/sipe-data/refs/heads/main/250304-W1D2.csv')
week1 = data1.to_numpy()
data2 = pd.read_csv('https://raw.githubusercontent.com/Andrew-Hamacher/sipe-data/refs/heads/main/250305-W1D3.csv')
week2 = data2.to_numpy()
data3 = pd.read_csv('https://raw.githubusercontent.com/Andrew-Hamacher/sipe-data/refs/heads/main/250306-W1D4.csv')
week3 = data3.to_numpy()

#input function for which spout you want to look at
spout = input('Which spout number?: ')
spout = int(spout)

#Plotting selected spout across the 3 days.
fig, ax = plt.subplots(3,1)
ax[0].plot(week1[:,0], week1[:,spout])
ax[0].set_title('Day 2')
ax[1].plot(week2[:,0], week2[:,spout])
ax[1].set_title('Day 3')
ax[2].plot(week3[:,0], week3[:,spout])
ax[2].set_title('Day 4')
plt.suptitle('Spout ' + str(spout) + ' across trials')
plt.tight_layout()
plt.show()
```
